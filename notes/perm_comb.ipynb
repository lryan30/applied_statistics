```python
# permutation of a set  is an arrangement of its elements into a sequence or linear order.
# a set containing n elements has n! permutations
# e.g. the set {A, B, C} has 3! = 6 permutations: ABC, ACB, BAC, BCA, CAB, CBA

# (3! # 3 factorial is 3*2*1 = 6)

# sets represent yes no order
# sets are unordered collections of distinct objects
# e.g. the set {A, B, C} is the same as the set {C, A, B}


# A combination is a selection of k elements from a set of size n, where the order of selection does not matter and can be selected twice.
# e.g. the set {A, B, C} has 3 combinations of size 2: AB, AC, BC

# k elements from a set of size n is given by the binomial coefficient C(n, k) = n! / (k! * (n - k)!)
# e.g. C(3, 2) = 3! / (2! * (3 - 2)!) = 3 / (2 * 1) = 3

# 3 choose 2 = 3! / (2! * (3 - 2)!) = 3 / (2 * 1) = 3

```
