math.comb is a built in function from the math module that calculates combinations, that is how many ways you can choose k items from n items without regard to order and without repititon. 



n --> the total number of items you have.
k --> the number of items you want to choose from those n.

Label the eight cups of tea `1`, `2`, `3`, `4`, `5`, `6`, `7`, and `8`.

We select four cups to have milk in first.

For example, we might select `2`, `4`, `5`, and `8`.

Visualize having four placeholders where we place four of the cup labels.

How many ways are there to select four cups from eight?

Cups: `1 2 3 4 5 6 7 8`

Placeholders: `_ _ _ _`

```python
# Total number of ways of selecting four cups from eight, keeping them in order.
ways_ordered = 8 * 7 * 6 * 5 # multiply the number of choices for each cup to get total number of ways 

# Show.
ways_ordered

# math module says its 70 different combinations but incorrect answer of 1680. 
#    for every first cup choice (8), there are 7 choices for second cup, 6 for third cup and 5 for fourth cup.
#    but this counts different orders of the same cups as different combinations.

# its counting permutations not combinations. 
# permutations = order matters
# combinations = order does not matter
# so we need to divide by the number of ways to order 4 cups (4!) to get combinations.

ways_corrected = ways_ordered // (4 * 3 * 2 * 1) # divide by 4! to correct for order
# Show.
ways_corrected

# permutations are greater as they include every possible ordering of the same cups
# combinations only count each unique set of cups once regardless of order.
```

```python
# No of ways of shuffling four cups.
no_shuffles = 4 * 3 * 2 * 1

# Show.
no_shuffles

# 24 different ways to shuffle four cups to place them in different orders. 
# Number of shuffles is the factorial of the number of cups (4! = 24).
```

```python
# No of combinations.
no_combs = ways_ordered // no_shuffles     # // integer division to get whole number. / float division would give decimal.

# Show.
no_combs
```
